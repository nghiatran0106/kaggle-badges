# Titanic model pipeline (badge baseline)
Trains a model and writes it to /kaggle/working so it can be saved as a Kaggle Model from the notebook output.

```python
import glob, pickle
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import cross_val_score

path = glob.glob("/kaggle/input/**/train.csv", recursive=True)[0]
train = pd.read_csv(path)
print(path, train.shape)
```

```python
FEATURES = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]

def prepare(df):
    x = df[FEATURES].copy()
    x["Sex"] = (x["Sex"] == "female").astype(int)
    x["Embarked"] = x["Embarked"].fillna("S").map({"S": 0, "C": 1, "Q": 2})
    return x.fillna({"Age": 28.0, "Fare": 14.45})

X, y = prepare(train), train["Survived"]
model = GradientBoostingClassifier(random_state=42)
print("CV acc", cross_val_score(model, X, y, cv=5).mean())
```

```python
model.fit(X, y)
with open("/kaggle/working/titanic_gbm.pkl", "wb") as f:
    pickle.dump(model, f)
print("saved")
```
