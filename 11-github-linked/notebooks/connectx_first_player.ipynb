Imported from GitHub (Github Coder badge).

```python
import pandas as pd
url = "https://raw.githubusercontent.com/nghiatran0106/kaggle-badges/main/11-github-linked/dataset/connectx_games.csv"
games = pd.read_csv(url)
games.groupby(["agent_1", "agent_2"]).winner_mark.value_counts(normalize=True).unstack()
```
