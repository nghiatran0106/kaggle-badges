# Titanic EDA in R (badge baseline)
Minimal R notebook: load, summarise, one plot.

```r
library(tidyverse)
path <- list.files("/kaggle/input", pattern = "^train.csv$", recursive = TRUE, full.names = TRUE)[1]
print(path)
train <- read_csv(path, show_col_types = FALSE)
glimpse(train)
```

```r
train %>% group_by(Sex, Pclass) %>% summarise(survival = mean(Survived), n = n(), .groups = "drop")
```

```r
ggplot(train, aes(Age, fill = factor(Survived))) +
  geom_histogram(bins = 30, na.rm = TRUE) +
  labs(fill = "Survived", title = "Age distribution by survival")
```
