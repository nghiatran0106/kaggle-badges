# ConnectX: how much is moving first worth?

A short look at 400 self-play games from the [ConnectX Self-Play Game Logs](https://www.kaggle.com/datasets/tranphunghia233871/connectx-self-play-game-logs) dataset: a simple heuristic agent against the built-in `random` and `negamax` agents.

```python
import glob
import os

import pandas as pd

games_path = glob.glob("/kaggle/input/**/games.csv", recursive=True)[0]
base = os.path.dirname(games_path)
games = pd.read_csv(games_path)
moves = pd.read_csv(os.path.join(base, "moves.csv"))
print(games.shape, moves.shape)
games.head()
```

## Win rate by pairing
`winner_mark` is 1 or 2 for the winning side and 0 for a draw.

```python
games.groupby(["agent_1", "agent_2"]).winner_mark.value_counts(normalize=True).unstack(fill_value=0).round(2)
```

## Does the opening column matter?
Win rate of the first player, grouped by the column of the very first move.

```python
first = moves[moves.move_no == 1][["game_id", "column"]].merge(games, on="game_id")
first["first_player_won"] = first.winner_mark == 1
first.groupby("column").first_player_won.agg(["mean", "count"]).round(2)
```
